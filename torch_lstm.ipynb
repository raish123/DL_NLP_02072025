```python
#importing module or library which will be used in this project.
import torch #this library we used to perform tensor computation
import torch.nn as nn
import numpy as np
import pandas as pd
from torch.utils.data import DataLoader,Dataset
from torch.nn.functional import softmax
```

```python
cleaned_line = []
with open('unstructure.txt','r') as file:
    data = file.read().splitlines()  # list of lines
    #removing empty space between multiple lines.
    for line in data:
        line = line.strip()  # removing whitespace from each line
        if line:  # if line empty nahi hai tabhi add karo
            cleaned_line.append(line)

cleaned_text = "\n".join(cleaned_line)
print(cleaned_text)
```

# now applying text preprocessing to textual data changing them into tokens

```python
import nltk
nltk.download('punkt_tab')
```

```python
import re
from nltk.tokenize import word_tokenize
def clean_text(text):
    #before changing the sentence to word tokens first of all we have to do cleaning.
    email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
    str_special = r'[^a-zA-Z0-9\s-]'
    url_pattern = r'http[s]?://(?:[a-zA-Z]|[0-9]|[$-_@.&+]|[!*\(\),]|(?:%[0-9a-fA-F][0-9a-fA-F]))+'
    html_pattern = r'<[^>]*>.*?</[^>]*>'
    
    #changing the senetence to lower case.
    text = text.lower()
    
    #applying regular expression pattern to sentence.
    text = re.sub(email_pattern,"",text)
    text = re.sub(str_special,"",text)
    text = re.sub(url_pattern,"",text)
    text = re.sub(html_pattern,"",text)
    print(text)
    
    
    lst_word = word_tokenize(text)
    return lst_word
```

```python
lst_word = clean_text(cleaned_text)
lst_word[:5]
```

# building vocubolary based on lst of token we have

```python
#creating empty dictatonary 
vocab = {'<unk>':0}

for word in lst_word:
    if word not in vocab:
        vocab[word] = len(vocab)
vocab
```

```python
#if i want to show how many no of unique vocobalory present in vocab object.
len(vocab)
```

# based on vocabolary we have build changing the sequential data to integer encoding

```python
def integer_encoding(sentence,vocab):
    text_encoding = []
    sentence = word_tokenize(sentence) #rtn lst of words
    for word in sentence:
        if word in vocab:
            text_encoding.append(vocab[word])
        else:
            text_encoding.append(vocab['<unk>'])
    return text_encoding
    
```

```python
#now iterating the sentence from cleaned_text and passing them integer_encoding to get integer sequence data.
input_numerical_sentences = []
for line in cleaned_text.split('\n'):
    input_numerical_sentences.append(integer_encoding(line,vocab))
```

```python
input_numerical_sentences
```

```python
#if i want to show max_length of sequential data present in input_numerical_sentences
max_length = max(len(sent) for sent in input_numerical_sentences)
max_length
```

# now applying prepadding to numerical_input_sentence so that all length we get to be same and preparing input and output data

```python
training_seq = []
for sent in input_numerical_sentences:
    for i in range(1,len(sent)):
        training_seq.append(sent[:i+1])
```

# doing prepadding to training_sequence

```python
#to show the maximum of sequential inputs data.
length_lst = []
for seq in training_seq:
    length_lst.append(len(seq))
max_len = max(length_lst)
max_len
    
```

```python
padded_sequence = []
for seq in training_seq:
    padded_sequence.append([0] * (max_len - len(seq)) + seq)
    
```

# converting unstructure data to supervised data

```python
input_data = padded_sequence[:,:-1]
output_data = padded_sequence[:,-1]
```

# creating custom dataset class

```python
class NextWordDataset(Dataset):
    #constructor method we used to initalize the instance variable.
    def __init__(self,input,output):
        super().__init__()
        self.input = input
        self.output = output
        
    
    def __len__(self):
        return len(self.input)
    
    def __getitem__(self, index):
        #this special method (__getitem__) based on index selecting input and output variable.
        return self.input[index],self.output[index]
  
```

```python
#creating an object of custom dataset class.
seq_data = NextWordDataset(input_data,output_data)
seq_data
```

# # using dataloader class we are loading data into batch_size.

```python
#creating an object of dataloader class
seq_dataloader = DataLoader(seq_data,batch_size=32,shuffle=True,pin_memory=True)
```

# # defining neural network architecture

```python
class SimpleLSTM(nn.Module):

    def __init__(self, vocab_size):
        #in this constructor method we are defining the architecture layer of lstm
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim=100)
        self.lstm = nn.LSTM(100,256,batch_first=True) #to get shape will [batch,seq_len,sequ_emb_vector_size]
        self.fc = nn.Linear(256, vocab_size)
    

    def forward(self,input_seq):
        #passing input to defined architecture layer of lstm
            embedded_vector = self.embedding(input_seq)
            intermediate_prev_hidden_state,(final_hidden_state,final_cell_state) = self.lstm(embedded_vector)
            output = self.fc(final_hidden_state.squeeze(0))

            return output
```

```python
#creating an object of Custom RNN Neural Network Architecture.

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

vocab_size = len(vocab)
model = SimpleLSTM(vocab_size=vocab_size).to(device)
model
```

```python
#if i want to see how many trainable parameter will be train by model
from torchinfo import summary
summary(model=model)
```

# now defining learning rate epochs and optimizer 

```python
learning_rate = 0.01
optimizers = torch.optim.Adam(model.parameters(),lr=learning_rate,weight_decay=0.001)
#optimizer will update weights and bias in each layer at time of training
epochs = 20

#now defining the loss function for multiclassifer and its inter encode so we are using sparse_categorical cross entropy
loss_fxn = torch.nn.CrossEntropyLoss()
loss_fxn
```

# now starting the training of model

```python
for epoch in range(epochs):
    model.train()
    training_loss_count = 0
    
    #first we loading the dataset then passing them into neural network architecture.
    for question,answer in seq_dataloader:
        question = question.to(device)
        answer   = answer.to(device)
        
        #passing input sequential to architecture first algo that work is forward pass geenrate the predicted output.
        prediction = model.forward(question) #prediction of output expected in this shape [batch_size,voc_size]
        #but we are getting [batch-size,seq_lenth,voc_size] so thatswhy we have done changes in last_layer.

        
        optimizers.zero_grad()
        
        #calculating the loss value
        losses = loss_fxn(prediction,answer)
        
        losses.backward()
        
        optimizers.step()
        
        training_loss_count = training_loss_count+losses
    print(f"epochs: {epoch+1} and train_loss: {training_loss_count}")
  
```

# predicting the model

```python
def prediction(model,vocab,text):
    
    #converting them into integer encoding
    integer_text = integer_encoding(text,vocab)
    
    #padding to text sequence
    padded_text = [0]*(max_length - len(integer_text)) + integer_text
    
    #converting them to tensor object
    padded_tensor = torch.tensor(padded_text,dtype=torch.long).unsqueeze(0).to(device) 
    
    #sending them to model
    logits = model(padded_tensor)
    
    #sending this logits to softmax function
    probs = torch.nn.functional.softmax(logits,dim=1)
    
    value,index = torch.max(probs,1)
    
    #merge with text
    return text + " " + list(vocab.keys())[index]
    
    
```
